```python
import os
import subprocess
import sys

REPO_URL = "https://github.com/selincyr/ah1s-rl-project.git"
REPO_DIR = "/content/ah1s-rl-project"

# Repo yoksa indir, varsa güncelle
if not os.path.exists(REPO_DIR):
    print("📥 Repo indiriliyor...")
    subprocess.run(
        ["git", "clone", REPO_URL, REPO_DIR],
        check=True
    )
else:
    print("🔄 Repo güncelleniyor...")
    subprocess.run(
        ["git", "-C", REPO_DIR, "pull"],
        check=True
    )

# Repo klasörüne geç
os.chdir(REPO_DIR)

# Gerekli paketleri kur
print("📦 Gerekli paketler kontrol ediliyor...")
subprocess.run(
    [sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"],
    check=True
)

print("\n✅ PROJE HAZIR")
print("Çalışma klasörü:", os.getcwd())
```
